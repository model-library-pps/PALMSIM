```python
# reported in Corley et al, 1971
# year - trunk mass

xys = [[1.5, 4.7],
        [2.5, 13.8],
        [4.5, 43.6],
        [6.5, 78.9],
        [8.5, 103.7],
        [10.5, 145.3],
        [14.5, 232.6],
        [17.5, 290.7],
        [27.5, 300.5],]
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from scipy.optimize import curve_fit
import numpy as np

def g(x,a,b,c):
    y = a*np.exp(-b*np.exp(-c*x))
    return y

params, _ = curve_fit(g, xs, ys)
ygs = [g(x,*params) for x in xs]
```

```python
xs = [xy[0] for xy in xys]
ys = [xy[1] for xy in xys]

plt.scatter(xs,ys)

xs = np.arange(0,30,1)
ygs = [g(x,*params) for x in xs]

plt.plot(xs,ygs)

dydxs = []

dx = 0.1

for x in xs:
    num = g(x+dx, *params) - g(x, *params)
    denum = dx
    dydx = num/denum
    dydxs += [dydx]
```

```python
[[x,round(y,1)] for x,y in zip(xs[::3], dydxs[::3])]
```
