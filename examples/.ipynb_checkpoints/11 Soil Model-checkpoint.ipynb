```python
import sys

sys.path.append('../PalmSim')
sys.path.append('..')

from PalmField import Soil

import pandas as pd
```

```python
s = Soil()
```

```python
s._rainfall_monthly = 50
```

```python
nmonths = 12

rainfall_series = [120,120,120,0,0,0,120,120,120,0,0,0,]

res = {}

for i in range(nmonths):
    
    rainfall = rainfall_series[i]
    
    s._rainfall_monthly = rainfall
    
    s.update()
    
    res[i] = s.to_dict()
    
df = pd.DataFrame(res).T
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

c = 'available_water (mm)'
xs = df[c].index
ys = df[c].values

plt.scatter(xs,ys)
plt.plot(xs,ys)
```
