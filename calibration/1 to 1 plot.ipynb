```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import os
```

```python
df = pd.read_csv('overview_results.csv', engine='python')
```

```python
list(df)
```

```python
sources = df['Source'].unique()
```

```python
for c in ['Avg. Yp', 'Avg. Yw']:

    fig, ax = plt.subplots(figsize=(4,4))

    lims = [0, 60]

    ax.axis('equal')

    for i, source in enumerate(sources):

        mask = df['Source'] == source

        xs = df.loc[mask , c].values
        ys = df.loc[mask,'Avg. Ya'].values

        print(source)

        ax.scatter(xs,ys, label='Company {:}'.format(i))

    ax.plot(lims,lims, color='black', linestyle='dashed')
    ax.legend()

    ax.set_xlim(lims)
    ax.set_ylim(lims)

    ax.set_xlabel('{:} (t/ha/yr)'.format(c))
    ax.set_ylabel('Avg. Ya (t/ha/yr)')
    
    plt.savefig('plots/1-to-1 {:}.png'.format(c),dpi=300)

```
