```python
w = 12
mask = df['window'] == w
```

```python
sdf = df.loc[mask].sort_values(by=['SD FFB Yw-Ya','SD BC Yw-Ya'])
```

```python
rank = [x+1 for x in range(len(sdf.index))]
```

```python
sdf['rank'] = rank
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def vsrank(c):
    xs = sdf['rank']
    ys = sdf[c]
    plt.scatter(xs,ys)
```

```python
sdf['SD FFB Ya - SD FFB Yw'] = sdf['SD FFB Ya'] - sdf['SD FFB Yw']
```

```python
vsrank('SD FFB Yw-Ya')
vsrank('SD FFB Ya - SD FFB Yw')
```
