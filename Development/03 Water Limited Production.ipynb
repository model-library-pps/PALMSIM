# Water Limited Production

In this example we use PalmSim to explore the **water limited production** for a water-limited site.

To be specific, we will consider a region of **Bangladesh**.

### What we expect

Given the water-limitation we expect drought and drought responses.

#### Drought responses

We believe that oil palm responds to **drought** primarily via

    1. Increased bunch abortion (bunch failure) --- around 2 months after anthesis.
    2. Increased inflorescence abortion (less bunch initiation) --- around 2 months before anthesis.
    3. Decrease in the female:male sex ratio at sex determination --- around 11 months after inflorescence initiation.
    
We expect these responses to result in a **decrease in bunch count** with the following time-lags.
 
     1. Around 4 months after the drought, related to bunch abortion.
     2. Around 8 months after the drought, related to inflorescence abortion.
     3. Around 30 months after the drought, related to sex differentiation.


### This Example   

This example has the following outline:

    1. We have a look at the input **weather data**.
    2. We consider the potential bunch yield (by dis-considering drought response).
    3. We consider the water-limited bunch yield (by considering drought response).
    4. We compare the two.
 

## Weather data

Most crop simulations require weather input data. Most commonly:
    - Solar irradiance (MJ/m2/day)
    - Precipation (mm/month)
    
And less important
    - Days with rain (1/month)
    
This demo comes with weather data files which can be found in the inputs folder.
    
**Note, these input files can for instance be opened/altered in Excel.**

````python
# import pandas --- spreadsheet library
import pandas as pd

# import matplotlib --- plotting library
import matplotlib.pyplot as plt 

# read-in the csv-file
weather_data = pd.read_csv('./input/bangladesh.csv',index_col='Date')
weather_data.index = pd.to_datetime(weather_data.index)
````

#### Note the quite variable solar radiation

````python
from plotting import tsplot, add_highlight

f,ax = tsplot(weather_data,'Solar (MJ/m2)')
````

#### Note the periodic drought!

````python
f,ax = tsplot(weather_data,'Precip (mm)')
````

## Potential Yield

Now let us **run PalmSim** given the **solar radiation data**.

First by taking the mean solar radiation as an input.


````python
import sys
sys.path.append('../PalmSim')
sys.path.append('..')
from PalmField import PalmField

year_of_planting = weather_data.index.year[0]

# 1. Initialize the model
p = PalmField(year_of_planting=year_of_planting)

# 2. Couple the solar radiation data:
p.weather.radiation_series = [weather_data['Solar (MJ/m2)'].mean()]

# 3. Run the model
df1 = p.run(duration=360,)
````

#### Viewing the results

Running PalmSim returns a spreadsheet which can be viewed directly and can be saved as an Excel spreadsheet (.csv).

Let us visualize the results directly.

To save the file such that it can be opened in e.g. excel simply run a piece of code

```python
df1.to_csv(filename)
```

````python
f,ax = tsplot(df1,'organs yield FM yearly (tonne FM/year)')
f,ax = tsplot(df1,'organs bunch count (1/ha/month)')
f,ax = tsplot(df1,'organs bunch weight (kg DM)')
````

#### Note the increase in bunch weigh over time!

There is also a peak in weight for the first bunches. This is not observed in reality and tells us that photosynthesis during the growth of these first bunches must have been less than expected! This agrees with the experimental proof that photosynthesis is relatively weak up to around the first real bunches!


## Potential Yield

Now let us **run PalmSim** given the actually variable solar radiation.

````python
# 1. Initialize the model
p = PalmField(year_of_planting=year_of_planting)

# 2. Couple weather data
p.weather.radiation_series = weather_data['Solar (MJ/m2)'].values

# 3. Run the model
df1 = p.run(duration=360,)
````

````python
f,ax = tsplot(df1,'organs yield FM yearly (tonne FM/year)')
f,ax = tsplot(df1,'organs bunch count (1/ha/month)')
f,ax = tsplot(df1,'organs bunch weight (kg DM)')
````

#### Note the variation in yield in response to the solar radiation.

Since we already work with monthly time-resolution and since not much is known about assimilate storage in palm, we did not model storage/release of assimilates which probably results in an over-estimation of the fluctuation of bunch weight. We are currently considering adding assimilate storage as an improvement for the next update -- note, this would smoothen the intra annual variations but the yearly production would remain the same.

## Water-Limited Yield

Just now we only considered solar radiation weather data.

Let us consider the drought response by coupling the precipitation data.

````python
# 1. Initialize the model
p = PalmField(year_of_planting=year_of_planting)

# 2. Couple weather data
p.weather.radiation_series = weather_data['Solar (MJ/m2)'].values
p.weather.rainfall_series = weather_data['Precip (mm)'].values # <-------------

# 3. Run the model
df2 = p.run(duration=360,)
````

````python
f,ax = tsplot(df2,'organs yield FM yearly (tonne FM/year)')
f,ax = tsplot(df2,'organs bunch count (1/ha/month)')
f,ax = tsplot(df2,'organs bunch weight (kg DM)')
````

Note the severe fluctuation in bunch count as a modelled response to drought!

# Comparison

Now let us simply compare both results:

````python
column = 'organs yield FM yearly (tonne FM/year)'
f,ax = tsplot(df1,column)
f,ax = tsplot(df2,column,ax=ax)
````

````python
column = 'organs bunch count (1/ha/month)'
f,ax = tsplot(df1,column)
f,ax = tsplot(df2,column,ax=ax)
````

````python
column = 'organs bunch weight (kg DM)'
f,ax = tsplot(df1,column)
f,ax = tsplot(df2,column,ax=ax)
````

What we observe is that the water-limited yield is significantly lower than the potential i.e. water is very limiting. Furthermore, the yield decrease is mainly via the bunch count (sex ratio, abortion), overall, bunch weight is calculated to remain approx. the same.
